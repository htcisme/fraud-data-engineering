```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

BASE_DIR = Path("../")

DATA_PATH = BASE_DIR / "data" / "raw" / "PS_20174392719_1491204439457_log.csv"

OUTPUT_DIR = BASE_DIR / "reports" / "figures"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(DATA_PATH, usecols=["type", "isFraud", "isFlaggedFraud"])
```

```python
type_count = df["type"].value_counts()
print(type_count)
# Type_count trả về kiểu object pandas series
# để truy xuất index ta dùng df["type"].index để lấy các key, item để truy xuất 1 item bên trong theo dạng tuple

type_percent = {
    k: v / type_count.sum() * 100 for k,v in type_count.items()
}

print(type_percent)
```

```python
# Tương tự với fraud count nè
fraud_counts = df["isFraud"].value_counts()
normal_counts = fraud_counts.get(0)
fraud_counts = fraud_counts.get(1)

print(normal_counts)

print(fraud_counts)

```
